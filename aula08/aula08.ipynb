```python
import random

numero_sorteado = random.randint(1, 50)

while True:
    numero_usuario = int(input("Digite um número: "))
    if numero_usuario == numero_sorteado:
        print("Parabéns!!")
        break
    else:
        print("errou!!")
        if numero_usuario > numero_sorteado:
            print("tente um numero menor")
        else:
            print("tente um numero maior!!")
```

```python
import random

numero_sorteado = random.randint(1, 50)
tentativas = 0  # contador de tentativas

while True:
    try:
        numero_usuario = int(input("Digite um número de 1 a 50: "))
    except ValueError:
        print("Por favor, digite apenas números inteiros!")
        continue  # volta pro começo do loop sem contar tentativa

    tentativas += 1  # aumenta o contador

    if numero_usuario == numero_sorteado:
        print(f"Parabéns!! Você acertou em {tentativas} tentativas!")
        break
    elif numero_usuario > numero_sorteado:
        print("Errou!! Tente um número menor.")
    else:
        print("Errou!! Tente um número maior.")
        
```
