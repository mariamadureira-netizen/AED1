Exercicío 1

```python
numero = int(input("Digite um número: "))

if numero % 5 == 0 and numero % 3 == 0:
    print("Divisível por 3 e por 5")
else:
    print("Não é divisível por 3 e por 5")
```

Exercicío 2

```python
valor = float(input("Digite o valor do produto: "))

if valor <= 1000:
    desconto = 0.10
elif valor <= 5000:
    desconto = 0.20
else:
    desconto = 0.30

valor_final = valor - (valor * desconto)

print("Valor final:", valor_final)
```
