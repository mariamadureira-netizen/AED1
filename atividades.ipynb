```python
livros = [
    {"titulo": "A Revolução dos Bichos", "autor": "George Orwell", "ano": 1945},
    {"titulo": "Dom Casmurro", "autor": "Machado de Assis", "ano": 1899},
    {"titulo": "O Hobbit", "autor": "J. R. R. Tolkien", "ano": 1937},
    {"titulo": "O Pequeno Príncipe", "autor": "Antoine de Saint-Exupéry", "ano": 1943},
    {"titulo": "1984", "autor": "George Orwell", "ano": 1949}
]

def busca_binaria_livros(livros, titulo):
    esquerda = 0
    direita = len(livros) - 1

    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        # Certifica-se que a comparação é feita de forma consistente (case-insensitive, se necessário)
        # Aqui, estamos assumindo que os títulos na lista e o título de busca são exatos (case-sensitive)
        if livros[meio]["titulo"] == titulo:
            return livros[meio]

        elif titulo < livros[meio]["titulo"]:
            direita = meio - 1

        else:
            esquerda = meio + 1

    return None

# PASSO 1: Garantir que a lista de livros esteja ordenada pelo título para que a busca binária funcione.
# A função sort é estável e eficiente.
livros.sort(key=lambda x: x['titulo'])

print("Lista de livros ordenada:")
for livro in livros:
    print(f"  - {livro['titulo']}")
print("\n")

# PASSO 2: Exemplo de uso da função de busca binária

# Busca por um livro existente
titulo_para_buscar_1 = "O Hobbit"
livro_encontrado_1 = busca_binaria_livros(livros, titulo_para_buscar_1)

if livro_encontrado_1:
    print(f"Livro encontrado '{titulo_para_buscar_1}': {livro_encontrado_1}")
else:
    print(f"Livro com o título '{titulo_para_buscar_1}' não encontrado.")

# Busca por um livro que não existe
titulo_para_buscar_2 = "O Senhor dos Anéis"
livro_encontrado_2 = busca_binaria_livros(livros, titulo_para_buscar_2)

if livro_encontrado_2:
    print(f"Livro encontrado '{titulo_para_buscar_2}': {livro_encontrado_2}")
else:
    print(f"Livro com o título '{titulo_para_buscar_2}' não encontrado.")

# Busca por outro livro existente
titulo_para_buscar_3 = "1984"
livro_encontrado_3 = busca_binaria_livros(livros, titulo_para_buscar_3)

if livro_encontrado_3:
    print(f"Livro encontrado '{titulo_para_buscar_3}': {livro_encontrado_3}")
else:
    print(f"Livro com o título '{titulo_para_buscar_3}' não encontrado.")
```

```python
catalogo_produtos = [
    {"nome": "Câmera DSLR", "preco": 2500.00, "descricao": "Câmera profissional com sensor full-frame."},
    {"nome": "Fone de Ouvido Bluetooth", "preco": 350.00, "descricao": "Fone sem fio com cancelamento de ruído."},
    {"nome": "Laptop Gamer", "preco": 7000.00, "descricao": "Notebook de alta performance para jogos."},
    {"nome": "Mouse sem Fio", "preco": 120.00, "descricao": "Mouse ergonômico com conexão sem fio."},
    {"nome": "Teclado Mecânico", "preco": 450.00, "descricao": "Teclado com switches mecânicos e retroiluminação."}
]

def busca_binaria_produto(catalogo, nome_produto):
    esquerda = 0
    direita = len(catalogo) - 1

    while esquerda <= direita:
        meio = (esquerda + direita) // 2
        nome_item = catalogo[meio]["nome"]

        if nome_item == nome_produto:
            return catalogo[meio]
        elif nome_produto < nome_item:
            direita = meio - 1
        else:
            esquerda = meio + 1
    return None

# PASSO 1: Garantir que o catálogo esteja ordenado alfabeticamente pelo nome do produto.
catalogo_produtos.sort(key=lambda x: x['nome'])

print("Catálogo de Produtos Ordenado:")
for produto in catalogo_produtos:
    print(f"  - {produto['nome']}")
print("\n")

# PASSO 2: Exemplos de uso da função de busca binária

# Buscar um produto existente
nome_para_buscar_1 = "Laptop Gamer"
produto_encontrado_1 = busca_binaria_produto(catalogo_produtos, nome_para_buscar_1)

if produto_encontrado_1:
    print(f"Produto encontrado '{nome_para_buscar_1}': {produto_encontrado_1}")
else:
    print(f"Produto com o nome '{nome_para_buscar_1}' não encontrado.")

# Buscar um produto que não existe
nome_para_buscar_2 = "Smart TV 4K"
produto_encontrado_2 = busca_binaria_produto(catalogo_produtos, nome_para_buscar_2)

if produto_encontrado_2:
    print(f"Produto encontrado '{nome_para_buscar_2}': {produto_encontrado_2}")
else:
    print(f"Produto com o nome '{nome_para_buscar_2}' não encontrado.")

# Buscar outro produto existente
nome_para_buscar_3 = "Fone de Ouvido Bluetooth"
produto_encontrado_3 = busca_binaria_produto(catalogo_produtos, nome_para_buscar_3)

if produto_encontrado_3:
    print(f"Produto encontrado '{nome_para_buscar_3}': {produto_encontrado_3}")
else:
    print(f"Produto com o nome '{nome_para_buscar_3}' não encontrado.")
```
