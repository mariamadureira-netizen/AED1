```python
# exercício 1 da lista

def calcular_animais (cabeças, pes):
  # p + c = cabeças
  # 2p + 4c = pes

  coelhos = (pes - 2 * cabeças ) // 2
  patos = cabeças - coelhos

  return patos, coelhos

  # entrada dos dados
cabeças = int(input("total de cabeças?(DOIS DIGITOS FINAIS DO RA)"))
pes = int(input("total de pes? ( cabeças * 3)"))

patos, coelhos = calcular_animais (cabeças, pes)

#saida
print("total de patos:", patos)
print("total de coelhos:", coelhos)
```

```python
# exercício 2 da lista

RA_input = input("digite seu RA:")
RA = int(RA_input)

soma = 0
multiplicaçao = 1

for n in RA_input:
  soma += int (n)
  multiplicaçao *= int (n)

print("RA:", RA)
print("soma:", soma)
print("multiplicação", multiplicaçao)
```

```python
# exercício 3 da lista

def reverso(numero):
  return int(str(numero)[::-1])


numero = int(input("digite um numero:"))

print("reverso", reverso(numero))
```
