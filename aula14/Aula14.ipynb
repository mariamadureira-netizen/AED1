```python
n = ["AC", "AP", "AM", "PA", "RO", "RR", "TO"]

ne = ["AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE"]

co = ["DF", "GO", "MT", "MS"]

se = ["ES", "MG", "RJ", "SP"]

s = ["PR", "RS", "SC"]


def regiao(r):

    if r in n or r in ne:
        return 50

    elif r in co or r in s:
        return 25

    elif r in se:
        return 0


def calcular_frete(peso, estado, presente=False):
"""
calcula o valor do frete baseado no peso,estado e se é ou não presente
entradas:
peso =  valor float em gramas 
estado = string com dois caracteres identificando o estado 
presente = apenas quando for presente, = true. valor padrao = false 
saidas:
valor em float do custo total do frete 
"""
    total = 0

    # peso
    if peso <= 500:
        total += 15

    elif peso <= 2000:
        total += 43

    else:
        total += 103

    # região
    total += regiao(estado)

    # presente
    if presente:
        total += 10

    return total


peso = float(input("Digite o peso em gramas: "))
estado = input("Digite a sigla do estado: ").upper()

resposta = input("É presente? (s/n): ").lower()

if resposta == "s":
    presente = True
else:
    presente = False


valor = calcular_frete(peso, estado, presente)

print("Valor do frete: R$", valor)
```

```python
import random

nome = input("Digite seu nome: ")

vj = 0
vc = 0
e = 0

while True:

    jogador = input("\nPedra, papel ou tesoura: ").lower()

    lista = ["pedra", "papel", "tesoura"]
    computador = random.choice(lista)

    print(nome, "jogou:", jogador)
    print("Computador jogou:", computador)

    if jogador == computador:
        print("Empate!")
        e += 1

    elif jogador == "pedra" and computador == "tesoura":
        print(nome, "venceu!")
        vj += 1

    elif jogador == "papel" and computador == "pedra":
        print(nome, "venceu!")
        vj += 1

    elif jogador == "tesoura" and computador == "papel":
        print(nome, "venceu!")
        vj += 1

    else:
        print("Computador venceu!")
        vc += 1

    continuar = input("\nQuer continuar? (s/n): ").lower()

    if continuar != "s":
        break


print("\nVITORIAS DE", nome, "......", vj)
print("VITORIAS DO COMPUTADOR......", vc)
print("EMPATES......", e)

if vj > vc:
    print(nome.upper(), "FOI O GRANDE CAMPEAO")

elif vc > vj:
    print("COMPUTADOR FOI O GRANDE CAMPEAO")

else:
    print("O JOGO TERMINOU EMPATADO")
```
